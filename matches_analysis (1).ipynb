```python
plt.figure(figsize=(12, 6))

team_wins.plot(kind="bar")

plt.title("IPL Matches Won by Team")
plt.xlabel("Team")
plt.ylabel("Number of Wins")
plt.xticks(rotation=90)
plt.show()
```

```python
top_players = df["player_of_match"].value_counts().head(10)

print(top_players)
```

```python
plt.figure(figsize=(10, 5))

top_players.plot(kind="bar")

plt.title("Top 10 Players by Player of the Match Awards")
plt.xlabel("Player")
plt.ylabel("Awards")
plt.xticks(rotation=45)
plt.show()
```

```python
top_venues = df["venue"].value_counts().head(10)

print(top_venues)
```

```python
plt.figure(figsize=(10, 6))

top_venues.plot(kind="barh")

plt.title("Top 10 IPL Venues by Number of Matches")
plt.xlabel("Number of Matches")
plt.ylabel("Venue")
plt.show()
```
